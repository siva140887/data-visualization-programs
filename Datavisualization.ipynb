```python
# Pandas series - It is a 1D array, labelled array. Each element (data) in this array will have a label (index). Elements can be of any data type. Index can be integers, strings, or timestamps
# Elements can be accessed using the integer position .iloc[] or other type position using .loc[]

import pandas as pd

# Creating Pandas series without index (default index will be integer starting from zero)
seriesdata = [21,22,23,24,25,26]
pdseries = pd.Series(seriesdata)
print(pdseries)
# Accessing 5th element using iloc and printing
print("5th element is:", pdseries.iloc[4])
# Finding the average of values in the series
average = pdseries.mean()
print("Average is", average)
# Finding max and min values and their indices
max1data = pdseries.max()
max1ind = pdseries.idxmax()
min1data = pdseries.min()
min1ind = pdseries.idxmin()
print("Max,min,indices:", max1data,max1ind,min1data,min1ind)
# Printing series with greater than a specific value
value1 = 22
print("New series:",pdseries[pdseries>value1])

# Creating Pandas series with string index
seriesdata = [21,22,23,24,25,26]
seriesindex = ['a','b','c','d','e','f']
pdseries = pd.Series(seriesdata, index=seriesindex)
print(pdseries)
# Accessing 6th element using loc and printing
print("6th element is:",pdseries.loc['f'])
# Finding max and min values and their indices
max1data = pdseries.max()
max1ind = pdseries.idxmax()
min1data = pdseries.min()
min1ind = pdseries.idxmin()
print("Max,min,indices:", max1data,max1ind,min1data,min1ind)
# Printing series with greater than a specific value
value1 = 22
print("New series:",pdseries[pdseries>value1])





```

```python
import pandas as pd
import numpy as np

###### 1
### creating dataframe of the dataset
df = pd.read_csv('/kaggle/input/cereals-dataset/cereal.csv')
print(df)
# print the first 5 rows
print(df.head())
# print 10 rows
print(df.head(10))

### picking numeric columns from dataframe
numericcolumns = df.select_dtypes(include=np.number).columns
print('The numeric columns are:',numericcolumns)

### preparing five number summary (min, Q1 (25%), Q2 (50%), Q3 (75%), max) using describe()
describeoutput = df[numericcolumns].describe()
print('Analysis output:')
print(describeoutput)

fivenosummary = describeoutput.loc[['min','25%','50%','75%','max']]
print('fivenosummary')
print(fivenosummary)

###### 2
### excluding the not mentioned numeric columns from the dataframe
df1 = df[numericcolumns].copy()  # creating a clone of df
print("before dropping:",df1)
df2 = df1.drop(['calories','shelf','weight','cups','rating'], axis=1)
print("after dropping:",df2)

numericcolumnsref = df2.columns
print(numericcolumnsref)

for col in numericcolumnsref:
  index = df2[col] == -1

  if index.any():   # if there is any index do the following
    meanval = df2.loc[~index,col].mean()  # while computing the mean ignore -1
    df2.loc[index,col] = meanval

print("Data frame after replacement:",df2)

###### 3
### preparing five number summary (min, Q1 (25%), Q2 (50%), Q3 (75%), max) using describe() after replacement
describeoutput1 = df2.describe()
print('Analysis output:')
print(describeoutput1)

fivenosummary1 = describeoutput1.loc[['min','25%','50%','75%','max']]
print('fivenosummary')
print(fivenosummary1)

###### 4

df3 = df2.copy()  #df3 is both missing and noisy values treated dataframe

### outliers are values that are either (< mean - 2 * std) or (> mean + 2 * std). Replacing them with attribute median

for col in df3.columns:
  meanval = df3[col].mean()
  stdval = df3[col].std()

  lowerbound = meanval - (2 * stdval)
  upperbound = meanval + (2 * stdval)

  index = (df3[col] < lowerbound ) | (df3[col] > upperbound)

  if index.any():
    medianval = df3[col].median()
    df3.loc[index,col] = medianval

### The noisy values treated data frame
print('noisy values treated data frame:', df3)

###### 5
### preparing five number summary (min, Q1 (25%), Q2 (50%), Q3 (75%), max) using describe() after replacement
describeoutput2 = df3.describe()
print('Analysis output:')
print(describeoutput2)

fivenosummary2 = describeoutput2.loc[['min','25%','50%','75%','max']]
print('fivenosummary')
print(fivenosummary2)


```
